```python
#concatenate in pandas
import pandas as pd
data1={
    'ID':['S01','S02','S03'],
    'Name':['Shreya','yashi','Manu'],
    'Roll':[11,12,13]
 
    }

data2={
    'ID':['S04','S05','S06'],
'Name':['Anjali','Veda','kashu'],
    'Roll':[14,15,16]
}


df1=pd.DataFrame(data1,index=['Student1','Student2','Student3'])
df2=pd.DataFrame(data2,index=['Student4','Student5','Student6'])
print("\nDataframe1:\n ",df1)
print("\nDataframe2:\n ",df2)

result=pd.concat([df1,df2])
print("\nConcatenation in two dataframes: \n",result)
```

```python
#Combine in pandas Series
import pandas as pd
data1=[10,30,50,70]
data2=[20,40,60,80]
df1=pd.Series(data1)
df2=pd.Series(data2)
print("Series Ist: \n",df1)
print("Series IInd: \n",df2)
def demo(x1,x2):
    if(x1>x2):
      return x1
    else:
      return x2

res=df1.combine(df2,demo)
print(res)
```

```python
#PANDAS-GROUP THE DATA
#1.group by:Split & combine the group
import pandas as pd

data = {
    "Player": [
        "Virat Kohli",
        "Rohit Sharma",
        "Virat Kohli",
        "Babar Azam",
        "Rohit Sharma",
        "Steve Smith",
        "Joe Root",
        "Babar Azam",
        "Kane Williamson",
        "Joe Root"
    ],
    "Rank": [1, 2, 1, 3, 2, 4, 5, 3, 6, 5],
    "Year": [2023, 2023, 2024, 2023, 2024, 2023, 2023, 2024, 2024, 2024]
}

df = pd.DataFrame(data)

print(df)
```

```python
res=df.groupby("Player")
print("\n",res.first())
```

```python
# Iterate through each group
for name, group in res:
    print("\nPlayer:", name)
    print(group)
```

```python
#3.View the group
print(df.groupby('Player').groups)
```

```python
#4.Get the mean of grouped Data
import pandas as pd
import numpy as np
data = {
    "Player": [
        "Virat Kohli",
        "Rohit Sharma",
        "Virat Kohli",
        "Babar Azam",
        "Rohit Sharma",
        "Steve Smith",
        "Joe Root",
        "Babar Azam",
        "Kane Williamson",
        "Joe Root"
    ],
    "Rank": [1, 2, 1, 3, 2, 4, 5, 3, 6, 5],
    "Year": [2023, 2023, 2024, 2023, 2024, 2023, 2023, 2024, 2024, 2024],
    "Points": [890, 850, 910, 830, 860, 780, 760, 845, 740, 775]
}

df = pd.DataFrame(data)

print(df)

# Group by Player
res = df.groupby("Year")

print(res["Points"].mean())
```

```python
#5.Get the size of each group
print("\n",res.agg(np.size))
```
